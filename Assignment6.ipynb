#  Scatter plot-
A scatter plot is a type of graph that displays individual data points on a two-dimensional plane. Each point represents a pair of values (x, y), showing how two variables relate to each other.


```python

import numpy as np
np.random.seed(3)
x = 3 + np.random.normal(0, 2, 50)
y = 3 + np.random.normal(0, 2, len(x))
```

```python
y
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(x=x,y=y)
plt.title('this is  my scatter plot')
plt.xlabel('this is my x axis')
plt.ylabel('this is my y axis')
```

# subplot()
The subplot() function in Matplotlib is used to create multiple plots within a single figure window. 
Instead of opening separate windows for each plot, you can organize several plots together in a grid layout.

```python

import numpy as np
x =np.array([0, 1, 2, 3, 4, 5]) 
y1 = np.array([0, 100, 200, 300, 400, 500])
x = np.array([0, 1, 2, 3, 4, 5]) 
y2= np.array([50, 20, 40, 20, 60, 70])
x = np.array([0, 1, 2, 3, 4, 5]) 
y3 = np.array([10, 20, 30, 40, 50, 60])
x = np.array([0, 1, 2, 3, 4, 5]) 
y4 = np.array([200, 350, 250, 550, 450, 150])
```

```python
# sub plot 1
plt.subplot(2,2,1)
plt.plot(x,y1,'r')

# sub plot 2
plt.subplot(2,2,2)
plt.plot(x,y2,'g')

# sub plot 3
plt.subplot(2,2,3)
plt.plot(x,y3,'b')

# sub plot 1
plt.subplot(2,2,4)
plt.plot(x,y4,'m')

plt.tight_layout()
plt.show()
```

# Q4: What is a bar plot? Why is it used? Using the following data plot a bar plot and a horizontal bar plot. 
 company = np.array(["Apple", "Microsoft", "Google", "AMD"])
 profit = np.array([3000, 8000, 1000, 10000])

# Bar plot-
A bar plot (or bar chart) is a type of graph that represents data using rectangular bars.

* The length/height of each bar corresponds to the value it represents.
* Bars can be drawn vertically or horizontally.
* Categories are usually shown on one axis, and numerical values on the other.
    
#  Why is it used-
1.Compare categories easily 
2.Show relative sizes
3.Visualize discrete data 
4.Communicate clearly



```python
import matplotlib.pyplot as plt
company = np.array(["Apple", "Microsoft", "Google", "AMD"])
profit = np.array([3000, 8000, 1000, 10000])

# vertical bar plot
plt.subplot(1,2,1)
plt.bar(company,profit,color='red')
plt.xlabel('company name')
plt.ylabel('profit')
plt.title('vertical bar plot')

# horizontal bar plot
plt.subplot(1,2,2)
plt.barh(company,profit,color='green')
plt.ylabel('company name')
plt.xlabel('profit')
plt.title('horizontal bar plot')
plt.tight_layout()
plt.show()
```
